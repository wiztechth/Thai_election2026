```python
import json
import pandas as pd

# 1. โหลดข้อมูลสถิติการเลือกตั้ง
with open('data/stats_cons.json', 'r', encoding='utf-8') as f:
    stats_cons = json.load(f)

# 2. ค้นหาข้อมูลเฉพาะจังหวัดชลบุรี (รหัส prov_id คือ 'CBI')
chonburi_data = next((prov for prov in stats_cons['result_province'] if prov['prov_id'] == 'CBI'), None)

if chonburi_data:
    summary_list = []
    # 3. วนลูปดึงข้อมูลรายเขตเลือกตั้ง
    for cons in chonburi_data.get('constituencies', []):
        summary_list.append({
            'เขตเลือกตั้ง': cons.get('cons_id'),
            'บัตรดี (ส.ส.เขต)': cons.get('valid_votes', 0),
            'บัตรเสีย (ส.ส.เขต)': cons.get('invalid_votes', 0),
            'ไม่ประสงค์ลงคะแนน (เขต)': cons.get('blank_votes', 0),
            'บัตรดี (บัญชีรายชื่อ)': cons.get('party_list_valid_votes', 0),
            'บัตรเสีย (บัญชีรายชื่อ)': cons.get('party_list_invalid_votes', 0),
            'ไม่ประสงค์ลงคะแนน (บัญชีรายชื่อ)': cons.get('party_list_blank_votes', 0)
        })

    # 4. แปลงข้อมูลเป็น DataFrame เพื่อความสวยงามและนำไปใช้งานต่อ
    df_cbi = pd.DataFrame(summary_list)

    # แสดงผลบนหน้าจอ
    print("สรุปคะแนนจังหวัดชลบุรี แยกตามเขตเลือกตั้ง:")
    print(df_cbi.to_string(index=False))

    # 5. บันทึกผลลัพธ์เป็นไฟล์ CSV
    df_cbi.to_csv('output/chonburi_vote_summary.csv', index=False, encoding='utf-8-sig')
    print("\nระบบบันทึกไฟล์ 'chonburi_vote_summary.csv' เรียบร้อยแล้ว")
else:
    print("ไม่พบข้อมูลของจังหวัดชลบุรีในไฟล์ที่ระบุ")
```
