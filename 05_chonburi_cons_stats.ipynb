```python
import json
import pandas as pd

# 1. Load the election statistics data
# Ensure 'stats_cons.json' is in the same directory
with open('data/stats_cons.json', 'r', encoding='utf-8') as f:
    stats_data = json.load(f)

# 2. Filter specifically for Chonburi province (ID: 'CBI')
chonburi_data = next((prov for prov in stats_data['result_province'] if prov['prov_id'] == 'CBI'), None)

summary_results = []

if chonburi_data:
    # Iterate through each constituency (district) in Chonburi
    for cons in chonburi_data.get('constituencies', []):
        cons_id = cons.get('cons_id')
        
        # --- Constituency (MP App) Data ---
        to_mp = cons.get('turn_out', 0)
        inv_mp = cons.get('invalid_votes', 0)
        blk_mp = cons.get('blank_votes', 0)
        
        # Calculate percentages
        inv_mp_pct = (inv_mp / to_mp * 100) if to_mp > 0 else 0
        blk_mp_pct = (blk_mp / to_mp * 100) if to_mp > 0 else 0
        
        # --- Party List Data ---
        to_pl = cons.get('party_list_turn_out', 0)
        inv_pl = cons.get('party_list_invalid_votes', 0)
        blk_pl = cons.get('party_list_blank_votes', 0)
        
        # Calculate percentages
        inv_pl_pct = (inv_pl / to_pl * 100) if to_pl > 0 else 0
        blk_pl_pct = (blk_pl / to_pl * 100) if to_pl > 0 else 0
        
        # --- Calculate Absolute Difference in Turnout (ABS Diff) ---
        abs_diff = abs(to_mp - to_pl)
        
        # Collect data for the table
        summary_results.append({
            'เขต': cons_id,
            'มาใช้สิทธิ (เขต)': to_mp,
            'บัตรเสีย (เขต)': inv_mp,
            'บัตรเสีย (เขต %)': round(inv_mp_pct, 2),
            'ไม่ประสงค์ (เขต)': blk_mp,
            'ไม่ประสงค์ (เขต %)': round(blk_mp_pct, 2),
            'มาใช้สิทธิ (บัญชี)': to_pl,
            'บัตรเสีย (บัญชี)': inv_pl,
            'บัตรเสีย (บัญชี %)': round(inv_pl_pct, 2),
            'ไม่ประสงค์ (บัญชี %)': round(blk_pl_pct, 2),
            'Diff แบบเขต-บัญชี': abs_diff
        })

    # 3. Create a DataFrame for clean formatting
    df = pd.DataFrame(summary_results)
    
    # Sort by District ID
    df = df.sort_values(by='เขต')

    # 4. Display the results
    print("--- สรุปสถิติการเลือกตั้งรายเขต จังหวัดชลบุรี ---")
    print(df.to_string(index=False))

    # Optional: Save the output to a CSV file
    # df.to_csv('output/chonburi_district_comparison.csv', index=False, encoding='utf-8-sig')
else:
    print("Error: Chonburi (CBI) data not found in the file.")
```
